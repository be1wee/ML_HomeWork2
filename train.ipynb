```python
import os, shutil

os.makedirs('/content/dataset/valid/images', exist_ok=True)
os.makedirs('/content/dataset/valid/labels', exist_ok=True)

train_img = '/content/dataset/train/images'
train_lbl = '/content/dataset/train/labels'

images = sorted(os.listdir(train_img))[:3]  

for img in images:
    name, _ = os.path.splitext(img)
    # Копируем изображение
    shutil.copy(os.path.join(train_img, img),
                os.path.join('/content/dataset/valid/images', img))
    label_file = name + '.txt'
    if os.path.exists(os.path.join(train_lbl, label_file)):
        shutil.copy(os.path.join(train_lbl, label_file),
                    os.path.join('/content/dataset/valid/labels', label_file))

print("Готово: папка valid создана и заполнена.")
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')

model.train(
    data='/content/dataset/data.yaml',
    epochs=100,
    imgsz=640,
    batch=8,
    project='/content/drive/MyDrive/yolo_training',
    name='car_number',
    exist_ok=True,
    device='cpu'
)
```
